```python
#loop control statement
```

```python
#break → Exits the loop.
#continue → Skips to the next iteration.
#pass → Does nothing (used as a placeholder).
```

```python
for i in range(5):
    if i == 3:
        break
    print(i)
```

```python
for i in range(5):
    if i == 3:
        continue
    print(i)
```

```python
for i in range(5):
    if i == 3:
        pass
    print(i)
```
